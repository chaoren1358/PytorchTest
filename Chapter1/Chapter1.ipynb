```python
import matplotlib.pyplot as plt
import numpy as np

# 数据
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# 创建图表
plt.plot(x, y)
print(type(x))
print(y)

# 添加标题和标签
plt.title("简单折线图",fontproperties='SimHei')
plt.xlabel("X 轴",fontproperties='SimHei')
plt.ylabel("Y 轴",fontproperties='SimHei')

# 显示图表
plt.show()

plt.plot(x, y, label='线性增长', color='blue', marker='o')
plt.legend()
plt.show()

categories = ['A', 'B', 'C', 'D']
values = [3, 7, 5, 9]

plt.bar(categories, values, color='green')
plt.title("柱状图")
plt.show()

data = np.random.randn(1000)
plt.hist(data, bins=30, color='purple', alpha=0.7)
plt.title("直方图")
plt.show()

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

axs[0, 0].plot(x, y)
axs[0, 0].set_title("折线图")

axs[0, 1].scatter(x, y)
axs[0, 1].set_title("散点图")

axs[1, 0].bar(categories, values)
axs[1, 0].set_title("柱状图")

axs[1, 1].hist(data, bins=20)
axs[1, 1].set_title("直方图")

plt.tight_layout()
plt.show()
```
